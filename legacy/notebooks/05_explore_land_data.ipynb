```python
import sys

sys.path.append("../src")

from load_data import load_canton_area, load_scenario_land, compute_relative_land

nc_path = "../data/raw/scenario_data/barrier_policy_alpine_pv,neighbours_fixed_by_EP2050_baseline_2016_result_4H.nc"

area_dict = load_canton_area("../data/raw/flache_der_schweiz_nach_kantonen_in_quadratkilometern.csv")
print(area_dict)

land = load_scenario_land(nc_path)
land_rel = compute_relative_land(land, area_dict)
print(land_rel[["canton", "tech", "land_use_km2", "relative_land"]].head(10))
```

```python
land_ch = land_rel[~land_rel["canton"].isin(["AUT", "DEU", "FRA", "ITA"])]
print(land_ch["relative_land"].describe())
print(land_ch.sort_values("relative_land", ascending=False).head(10))
```

```python
nc_path_wind = "../data/raw/scenario_data/barrier_policy_wind_x2,neighbours_fixed_by_EP2050_baseline_2016_result_4H.nc"

land_wind = load_scenario_land(nc_path_wind)
land_wind_rel = compute_relative_land(land_wind, area_dict)
land_wind_ch = land_wind_rel[~land_wind_rel["canton"].isin(["AUT", "DEU", "FRA", "ITA"])]

print(land_wind_ch["relative_land"].describe())
print(land_wind_ch.sort_values("relative_land", ascending=False).head(10))
```
